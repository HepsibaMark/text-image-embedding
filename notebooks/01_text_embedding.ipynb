```python
model_name = "openai/clip-vit-base-patch32"

tokenizer = CLIPTokenizer.from_pretrained(model_name)
text_model = CLIPTextModel.from_pretrained(model_name)
```

```python
text = "A cute puppy sitting in a garden"

inputs = tokenizer(
    text,
    return_tensors="pt",
    padding=True,
    truncation=True
)

print(inputs)
```

```python
with torch.no_grad():
    outputs = text_model(**inputs)

text_embedding = outputs.pooler_output

print("Embedding shape:", text_embedding.shape)
```
